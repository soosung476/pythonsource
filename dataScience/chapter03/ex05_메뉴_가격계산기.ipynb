# 배달 메뉴 가격 계산기 - Chapter02 데이터 재활용
- `gr.Dropdown`으로 메뉴를 선택하고 `gr.Slider`로 수량을 조절하는 인터페이스 생성
- Chapter02의 배달 메뉴 데이터를 활용

```python
import gradio as gr

menu_prices = {
    "떡볶이": 9000,
    "치킨": 22000,
    "피자": 25000,
    "짜장면": 8000,
    "마라탕": 13000,
    "돈까스": 11000,
}

def calculate_total(menu, quantity):
    unit_price = menu_prices[menu]
    total = unit_price* quantity
    return f"{menu} {quantity}개 주문 -> 총 결제 금액 : {total}원"
```

```python
demo = gr.Interface(
    fn=calculate_total,
    inputs = [
        # Dropdown : 메뉴 딕셔너리의 key들을 선택지로 사용
        gr.Dropdown(list(menu_prices.keys()),label="메뉴 선택"),
        # Slider : 1~10 사이 값을 슬라이더로 조절 (step=1: 1씩 증가 버튼)
        gr.Slider(minimum=1,maximum=10,step=1,label="수량"),
    ],
    outputs=gr.Textbox(label="주문요약"),
    title="배달 메뉴 가격 계산기"
)
demo.launch()
```

```python
demo.close()
```
